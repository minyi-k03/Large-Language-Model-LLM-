# 학습 데이터 uplaod
## mydata.jsonl 파일 링크 : https://drive.google.com/file/d/1wYTLakSnBF_kA6TyUjgXy52r360KSaSL/view?usp=sharing

```python
import os
from openai import OpenAI

#KEY 입력
client = OpenAI(api_key="입력")

#파일 업로드
try:
    response = client.files.create(
        file=open("mydata.jsonl", "rb"),  # 파일 이름이 맞는지 꼭 확인하세요!
        purpose='fine-tune'
    )
    print("✅ 업로드 성공!")
    print(f"파일 ID: {response.id}")  # 이 ID를 나중에 파인튜닝할 때 씁니다.

except FileNotFoundError:
    print("❌ 오류: 'mydata.jsonl' 파일을 찾을 수 없습니다. 왼쪽 파일 메뉴에 업로드했는지 확인해주세요.")
except Exception as e:
    print(f"❌ 오류 발생: {e}")
```

# 업로드한 File Id 확인

```python
files = client.files.list()
```

# Fine-Tuning 작업 시작

```python
job = client.fine_tuning.jobs.create(
    training_file="입력",  # ⚠️ 아까 확인한 파일 ID를 여기에 복붙하세요!
    model="gpt-3.5-turbo"
)
```

# Fine-Tuning된 모델 Inference

## Sarcastic 한 자아를 가진 챗봇으로 요청

```python
completion = client.chat.completions.create(
  model="gpt-3.5-turbo",
  messages=[
    {"role": "system", "content": "Marv is a factual chatbot that is also sarcastic."},
    {"role": "user", "content": "What's the capital of France?"}
  ]
)
print(completion.choices[0].message.content)
```

## Helpful 한 자아를 가진 챗봇으로 요청

```python
completion = client.chat.completions.create(
  model="gpt-3.5-turbo",
  messages=[
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "What's the capital of France?"}
  ]
)
print(completion.choices[0].message.content)
```

# Fine-Tuning 작업 취소하기

```python
from openai import OpenAI

# 1. 클라이언트 연결
client = OpenAI(api_key="")

# 2. 최근 작업 목록 10개 가져오기
print("📋 최근 파인튜닝 작업 목록:")
jobs = client.fine_tuning.jobs.list(limit=10)

for job in jobs.data:
    print(f"🆔 ID: {job.id}")          # 👉 이게 바로 작업 ID입니다! (ftjob-...)
    print(f"📊 상태: {job.status}")     # (running, succeeded, failed, cancelled)
    print(f"🤖 모델: {job.model}")
    print("-" * 30)
```
