# Quick start

```python
from openai import OpenAI
#API Key 설정
OPENAI_KEY = "Input Your Key"

#클라이언트 연결
client = OpenAI(api_key=OPENAI_KEY)
```

```python
response = client.responses.create(
    model="gpt-4o-mini",
    tools=[{"type": "web_search_preview"}],
    input="오늘 있었던 긍정적인 뉴스는 무엇인가요?"
)

print(response.output_text)
```

```python
response
```

```python
for item in response.output:
    if item.type == "message":
        for content in item.content:
            if content.type == "output_text":
                text = content.text
                annotations = getattr(content, "annotations", [])

                print("📅 오늘의 긍정적인 뉴스 (2025년 3월 18일)\n")

                for i, annotation in enumerate(annotations, 1):
                    title = annotation.title
                    url = annotation.url
                    summary = text[annotation.start_index:annotation.end_index].strip()

                    print(f"{i}. 📰 {title}")
                    print(f"   {summary}")
                    print(f"   🔗 기사 보기: {url}\n")
```

```python
from langchain_community.tools import TavilySearchResults

tool = TavilySearchResults(
    max_results=5
)
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0,
)
```

##LangChain과 Tavily를 이용한 답변 ##

```python
import datetime

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableConfig, chain

today = datetime.datetime.today().strftime("%D")
prompt = ChatPromptTemplate(
    [
        ("system", f"You are a helpful assistant. The date today is {today}."),
        ("human", "{user_input}"),
        ("placeholder", "{messages}"),
    ]
)

# specifying tool_choice will force the model to call this tool.
llm_with_tools = llm.bind_tools([tool])

llm_chain = prompt | llm_with_tools


@chain
def tool_chain(user_input: str, config: RunnableConfig):
    input_ = {"user_input": user_input}
    ai_msg = llm_chain.invoke(input_, config=config)
    tool_msgs = tool.batch(ai_msg.tool_calls, config=config)
    return llm_chain.invoke({**input_, "messages": [ai_msg, *tool_msgs]}, config=config)


tool_chain.invoke("오늘 있었던 긍정적인 뉴스는 무엇인가요?")
```

## tool_choice="required"로 변경

```python
import datetime

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableConfig, chain

today = datetime.datetime.today().strftime("%D")
prompt = ChatPromptTemplate(
    [
        ("system", f"You are a helpful assistant. The date today is {today}."),
        ("human", "{user_input}"),
        ("placeholder", "{messages}"),
    ]
)

# specifying tool_choice will force the model to call this tool.
llm_with_tools = llm.bind_tools([tool], tool_choice="required")

llm_chain = prompt | llm_with_tools


@chain
def tool_chain(user_input: str, config: RunnableConfig):
    input_ = {"user_input": user_input}
    ai_msg = llm_chain.invoke(input_, config=config)
    tool_msgs = tool.batch(ai_msg.tool_calls, config=config)
    return llm_chain.invoke({**input_, "messages": [ai_msg, *tool_msgs]}, config=config)
```

```python
tool_chain.invoke("2010년에 IEEE 프랭크 로젠블랫 상(Frank Rosenblatt Award)을 수상한 사람은 누구인가요?")
```

```python
tool_chain.invoke("Who received the IEEE Frank Rosenblatt Award in 2010?")
```

```python
tool_chain.invoke("How much money, in euros, was the surgeon held responsible for Stella Obasanjo's death ordered to pay her son?")
```
