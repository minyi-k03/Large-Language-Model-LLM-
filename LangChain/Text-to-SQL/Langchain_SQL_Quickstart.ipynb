# chinook.db 불러오기

````python
import os
from langchain_community.utilities import SQLDatabase

db_filename = "chinook.db"


db = SQLDatabase.from_uri(f"sqlite:///{db_filename}")

# 4. 연결 테스트 (employees 테이블이 보여야 정상)
print(f"테이블 목록: {db.get_table_names()}")
````

#모델 및 체인 설정

````python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


llm = ChatOpenAI(model="gpt-4o-mini", temperature=0, openai_api_key=OPENAI_KEY)


template = """
당신은 SQL 전문가입니다. 아래 스키마를 보고 질문에 맞는 쿼리를 작성해주세요.
오직 SQL 쿼리문만 출력하세요. (마크다운, 설명 금지)

[스키마 정보]
{schema}

[질문]
{question}
"""

prompt = ChatPromptTemplate.from_template(template)


chain = prompt | llm | StrOutputParser()
````

````python
response = chain.invoke({
    "schema": db.get_table_info(),
    "question": "employees 테이블에 직원이 몇 명 있어?"
})


clean_sql = response.replace("```sql", "").replace("```", "").strip()
print(f"Generated SQL: {clean_sql}")

````

# Execute SQL query

## 이 chain은 **실제로 쿼리를 DB에 실행하기 때문에 매우 주의해서 사용**해야합니다.

````python
from langchain_community.tools.sql_database.tool import QuerySQLDataBaseTool
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_community.utilities import SQLDatabase
from langchain_openai import ChatOpenAI


db = SQLDatabase.from_uri("sqlite:///chinook.db")
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0, openai_api_key=OPENAI_KEY)

# 2. [수정됨] write_query: SQL 생성 체인 (오류 나는 라이브러리 함수 대신 직접 정의)
template = """Answer the user's question by creating a syntactically correct SQLite query.
Only return the SQL query. Do not add any markdown, explanation, or backticks.

[Schema]
{schema}

[Question]
{question}
"""
prompt = ChatPromptTemplate.from_template(template)


write_query = prompt | llm | StrOutputParser()


execute_query = QuerySQLDataBaseTool(db=db)


def clean_sql(text):
    return text.replace("```sql", "").replace("```", "").strip()

chain = write_query | clean_sql | execute_query

# 5. 실행
response = chain.invoke({
    "schema": db.get_table_info(),
    "question": "employees 테이블에 데이터가 몇 개 있어?"
})

print(f"------------\n[실행 결과]\n{response}\n------------")
````

# Answer the question

````python
from operator import itemgetter
from langchain_core.prompts import ChatPromptTemplate, PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
from langchain_community.tools.sql_database.tool import QuerySQLDataBaseTool


def get_schema(_):
    return db.get_table_info()


sql_template = """You are a SQLite expert. Create a syntactically correct SQLite query to answer the user's question.
Only return the SQL query. No markdown, no explanation.

[Schema]
{schema}

[Question]
{question}
"""
sql_prompt = ChatPromptTemplate.from_template(sql_template)

write_query = (
    RunnablePassthrough.assign(schema=get_schema)
    | sql_prompt
    | llm
    | StrOutputParser()
)


execute_query = QuerySQLDataBaseTool(db=db)

# 마크다운 제거 함수
clean_sql_func = lambda x: x.replace("```sql", "").replace("```", "").strip()


answer_prompt = PromptTemplate.from_template(
    """Given the following user question, corresponding SQL query, and SQL result, answer the user question naturally in Korean.

Question: {question}
SQL Query: {query}
SQL Result: {result}
Answer: """
)

chain = (
    RunnablePassthrough.assign(query=write_query | clean_sql_func) # 1. SQL 생성
    .assign(
        result=itemgetter("query") | execute_query                 # 2. SQL 실행 (결과 획득)
    )
    | answer_prompt                                                # 3. 답변 프롬프트 입력
    | llm                                                          # 4. 최종 답변 생성
    | StrOutputParser()
)


#Execution
print("--------------------------------------------------")
response = chain.invoke({"question": "employees 테이블에 직원이 몇 명 있어?"})
print(f"최종 답변: {response}")
print("--------------------------------------------------")
````

# Agents

SQL Agent 는 SQL을 좀더 유연게 처리할 수 있도록 도와줍니다. SQL Agent의 장점은 아래와 같습니다.
*   데이터베이스의 스키마뿐만 아니라 데이터베이스의 내용을 바탕으로 한 질문에도 답할 수 있습니다
*   (특정 테이블을 설명하는 것과 같은). 생성된 쿼리를 실행하여 오류를 복구하고, 추적하여 정확하게 다시 생성할 수 있습니다.
* 여러 의존 쿼리가 필요한 질문에 대한 답변이 가능합니다.
* 관련 테이블의 스키마만 고려하여 토큰을 절약할 수 있습니다.






````python
#개별로 체인연결이 아닌 좀 더 유연하게 사용
from langchain_community.agent_toolkits import create_sql_agent

agent_executor = create_sql_agent(llm, db=db, agent_type="openai-tools", verbose=True)
````

````python
agent_executor.invoke(
    {
        "input": "국가별 총 판매액을 나열하세요. 어느 국가의 고객이 가장 많은 돈을 지출했나요?"
    }
)
````

````python
agent_executor.invoke(
    {
        "input": "가장 많은 앨범을 발매한 아티스트는 누구인가요?"
    }
)
````
